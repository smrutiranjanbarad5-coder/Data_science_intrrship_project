```python
import pandas as pd

df = pd.read_csv("tmdb_5000_movies.csv")

print(df.shape)
df.head()
```

```python
# Create tags for each movie

df['tags'] = (
    df['title'].fillna('') + ' ' +
    df['overview'].fillna('') + ' ' +
    df['genres'].fillna('') + ' ' +
    df['keywords'].fillna('')
)

print("Tags created successfully!")
print(df[['title', 'tags']].head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF
tfidf = TfidfVectorizer(stop_words='english')

# Convert tags into numbers
tfidf_matrix = tfidf.fit_transform(df['tags'])

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Create TF-IDF Vectorizer
tfidf = TfidfVectorizer(stop_words='english')

# Convert movie tags into numbers
tfidf_matrix = tfidf.fit_transform(df['tags'])

print("TF-IDF Matrix Shape:", tfidf_matrix.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

# Calculate similarity between all movies
similarity = cosine_similarity(tfidf_matrix)

print("Similarity Matrix Shape:", similarity.shape)
```

```python
# Movie Recommendation Function

def recommend(movie):
    
    # Find the movie index
    movie_index = df[df['title'].str.lower() == movie.lower()].index
    
    if len(movie_index) == 0:
        print("Movie not found!")
        return
    
    movie_index = movie_index[0]
    
    # Get similarity scores
    distances = similarity[movie_index]
    
    # Sort movies by similarity
    movie_list = sorted(
        list(enumerate(distances)),
        reverse=True,
        key=lambda x: x[1]
    )[1:6]
    
    print("Movies recommended for:", df.iloc[movie_index]['title'])
    print("--------------------------------")
    
    for i in movie_list:
        print(df.iloc[i[0]]['title'])
```

```python
recommend("Avatar")
```

```python
recommend("The Dark Knight Rises")
```
